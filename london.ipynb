```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("asaniczka/top-spotify-songs-in-73-countries-daily-updated")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# List files in the downloaded dataset path
files = os.listdir(path)
print("Files in dataset path:", files)

# Identify CSV files and load the first one into a DataFrame
csv_files = [f for f in files if f.endswith('.csv')]
if csv_files:
    csv_path = os.path.join(path, csv_files[0])
    df = pd.read_csv(csv_path)
    print(f"Loaded {csv_files[0]} successfully!")
    display(df.head())
else:
    print("No CSV files found in the dataset directory.")
```

```python
import pandas as pd

# categorical data
df = pd.DataFrame(
    {"song names": ["Ordinary", "Been That", "KYKY2BONDY", "Midnight Drive", "Paris",
                    "ZAAZAA", "Lotje (Lil Kleine Remix)", "Lotje", "Echte Liefde Is Te Koop",
                    "Stolt (feat. Lamin)", "Bipolar", "En drøm om et menneske", "Hørt Det Før (feat. Pil, Annika, Mille & Medina)", "Smelter Under Månen",
                    "99 (feat. Daecolm)", "Very Soon", "Pressure", "lost", "Al- Jannah",
                    "no tiene sentido", "Who", "La Plena - W Sound 05", "Don't Say You Love Me", "mi refe",
                    "LAW NASYANY", "بعبع دايرتى", "محكمه ودخلنا علي المفرمه - افتري و عوجلي شعره لورا", "كاتم الانفاس - محطوط تشكيل", "انا لبسي اوفر سايز" ]}
)

# categorical groups
group_mapping = {
    "Pop Umbrella": ["Ordinary", "Lotje", "Echte Liefde Is Te Koop", "Stolt (feat. Lamin)", "Hørt Det Før (feat. Pil, Annika, Mille & Medina)", "Smelter Under Månen", "Very Soon", "Pressure", "lost", "Al- Jannah", "La Plena - W Sound 05", "Don't Say You Love Me", "mi refe"],
    "Rock": ["Lotje (Lil Kleine Remix)", "En drøm om et menneske"],
    "Hip Hop/Rap": ["Been That", "KYKY2BONDY", "Paris", "ZAAZAA", "Bipolar", "Who", "LAW NASYANY"],
    "Electronic": ["Midnight Drive", "99 (feat. Daecolm)", "بعبع دايرتى", "محكمه ودخلنا علي المفرمه - افتري و عوجلي شعره لورا", "كاتم الانفاس - محطوط تشكيل", "انا لبسي اوفر سايز"],
    "Other":["no tiene sentido"]
}


inv_mapping = {item: group for group, items in group_mapping.items() for item in items}

#create a new categorical group column
df["song_genre"] = df["song names"].map(inv_mapping)

print(df)
```

```python

import pandas as pd

#FIX THE DATAFRAME TO INCLUDE THE AMOUNT OF TIMES EACH SONG IS INCLUDED IN THE ORIGINAL DATA

df=pd.Dataframe({

   'Music Genres' =['Pop Umbrella', 'Rock', 'Hip Hop/Rap', 'Electronic', 'Other']

})

counts=df ['Music Genres'].valuecounts()

print(counts)
```

```python
#graph
import matplotlib.pyplot as plt

# fix y by tallying the amount of data points and manually inputing

x1= [1]
y1= [Pop Umbrella],

x2= [2]
y2= [Rock],

x3= [3]
y3= [Hip Hop/Rap],

x4= [4]
y4= [Electronic],

x5= [5]
y5= [Other]

plt.bar(x1,y1, label= "Pop Umbrella", color='p')
plt.bar(x2,y2, label= "Rock", color='b')
plt.bar(x3,y3, label= "Hip Hop/Rap", color='o')
plt.bar(x4,y4, label="Electronic", color='g')
plt.bar(x5,y5, label="Other", color='y')
plt.plot()

plt.xlabel("Music Genre")
plt.ylabel ("Number of Times Recorded")
plt.title ("Is The Popularity of Music Genres Affected by a Country's Quality of Life?")
plt.legend()
plt.show()
```
